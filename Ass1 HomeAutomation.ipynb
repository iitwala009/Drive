```python
while True:

    print("\n--- HOME AUTOMATION SYSTEM ---")
    print("1. Automatic Light")
    print("2. Air Conditioner")
    print("3. Door Security")
    print("4. Water Tank")
    print("5. TV Control")
    print("6. Exit System")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        darkness = input("Is the room dark? (yes/no): ")

        if darkness == "yes":
            print("Light is turned ON")
        else:
            print("Light is turned OFF")

    elif choice == 2:
        temperature = int(input("Enter room temperature: "))

        if temperature >= 30:
            print("AC is turned ON")
        elif temperature <= 22:
            print("AC is turned OFF")
        else:
            print("AC remains in normal mode")

    elif choice == 3:
        person = input("Is an unauthorized person detected? (yes/no): ")

        if person == "yes":
            print("Door is LOCKED")
            print("Security Alarm is ON")
        else:
            print("Door is UNLOCKED")

    elif choice == 4:
        water = int(input("Enter water tank level (%): "))

        if water <= 20:
            print("Water level is low")
            print("Water Pump is ON")
        elif water >= 90:
            print("Water tank is full")
            print("Water Pump is OFF")
        else:
            print("Water level is normal")

    elif choice == 5:
        tv = input("Do you want to watch TV? (yes/no): ")

        if tv == "yes":
            print("TV is turned ON")
        else:
            print("TV is turned OFF")

    elif choice == 6:
        print("Home Automation System stopped")
        break

    else:
        print("Invalid choice. Please enter a valid option.")
```
