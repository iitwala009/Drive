```python
# Display Troubleshooting Expert System

knowledge_base = [
    {
        "problem": "Display is powered OFF",
        "conditions": {"display_power_off": True},
        "solution": "Check the power cable and switch ON the display."
    },
    {
        "problem": "Display cable connection problem",
        "conditions": {
            "display_power_off": False,
            "no_display_signal": True
        },
        "solution": "Reconnect the HDMI or DisplayPort cable."
    },
    {
        "problem": "Incorrect display settings",
        "conditions": {
            "display_power_off": False,
            "no_display_signal": False,
            "display_settings_wrong": True
        },
        "solution": "Set a supported screen resolution and refresh rate."
    },
    {
        "problem": "Display driver problem",
        "conditions": {
            "display_power_off": False,
            "no_display_signal": False,
            "display_settings_wrong": False,
            "display_driver_issue": True
        },
        "solution": "Update or reinstall the graphics driver."
    }
]

questions = {
    "display_power_off": "Is the display power OFF?",
    "no_display_signal": "Does the monitor show No Signal?",
    "display_settings_wrong": "Are the display settings incorrect?",
    "display_driver_issue": "Is a display driver problem suspected?"
}


def get_answer(question):
    while True:
        answer = input(question + " (yes/no): ").lower()

        if answer == "yes" or answer == "y":
            return True
        elif answer == "no" or answer == "n":
            return False
        else:
            print("Invalid input. Enter yes or no.")


# Forward Chaining

def forward_chaining():
    print("\n===== FORWARD CHAINING =====")
    print("Answer the following questions:\n")

    facts = {}

    for rule in knowledge_base:
        for fact in rule["conditions"]:
            if fact not in facts:
                facts[fact] = get_answer(questions[fact])

    for rule in knowledge_base:
        matched = True

        for fact, expected_value in rule["conditions"].items():
            if facts[fact] != expected_value:
                matched = False
                break

        if matched:
            print("\nProblem identified:", rule["problem"])
            print("Recommended solution:", rule["solution"])
            return

    print("\nNo matching rule found.")
    print("Check the display hardware, connections, and drivers.")


# Backward Chaining

def backward_chaining():
    print("\n===== BACKWARD CHAINING =====")

    facts = {}

    for rule in knowledge_base:
        print("\nChecking goal:", rule["problem"])

        goal_proved = True

        for fact, expected_value in rule["conditions"].items():

            if fact not in facts:
                facts[fact] = get_answer(questions[fact])

            if facts[fact] != expected_value:
                goal_proved = False
                print("Condition not satisfied:", fact)
                break

            print("Condition satisfied:", fact)

        if goal_proved:
            print("\nProblem confirmed:", rule["problem"])
            print("Recommended solution:", rule["solution"])
            return

    print("\nNo problem could be confirmed using the available rules.")
    print("Check the display connections, settings, and graphics driver.")


# Main Program

def display_troubleshooting_system():

    while True:
        print("\n================================")
        print("DISPLAY TROUBLESHOOTING EXPERT SYSTEM")
        print("================================")
        print("1. Forward Chaining")
        print("2. Backward Chaining")
        print("3. Exit")

        choice = input("Enter your choice (1-3): ")

        if choice == "1":
            forward_chaining()

        elif choice == "2":
            backward_chaining()

        elif choice == "3":
            print("Exiting Display Troubleshooting System.")
            break

        else:
            print("Invalid choice. Please enter 1, 2, or 3.")


display_troubleshooting_system()
```
