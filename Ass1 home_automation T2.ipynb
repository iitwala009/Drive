```python
def home_automation(temperature, person_present, room_dark):

    if temperature > 28:
        fan = "ON"
    else:
        fan = "OFF"

    if temperature > 28 and person_present:
        ac = "ON"
    else:
        ac = "OFF"

    if room_dark and person_present:
        light = "ON"
    else:
        light = "OFF"

    print("=======Input==========")
    print("Temp : ", temperature)
    print("person : ", person_present)
    print("Room dark : ", room_dark)

    print("========Prediction======")
    print("Fan", fan)
    print("AC", ac)
    print("Light", light)


automation = True

while automation:

    print("==========Menu==========")
    print("1. Run automation")
    print("2. Exit")

    ch = int(input("Choose: "))

    if ch == 1:

        while True:
            temperature_input = input("Temperature (in °C): ")

            valid_temp = temperature_input.replace(".", "", 1).replace("-", "", 1).isdigit()

            if valid_temp:
                temperature = float(temperature_input)
                break
            else:
                print("Invalid Temp")

        while True:
            person = input("Person (yes/no): ").lower()

            if person == "yes":
                person_present = True
                break
            elif person == "no":
                person_present = False
                break
            else:
                print("Invalid input")

        while True:
            Dark = input("Room dark (yes/no): ").lower()

            if Dark == "yes":
                room_dark = True
                break
            elif Dark == "no":
                room_dark = False
                break
            else:
                print("Invalid input")

        home_automation(temperature, person_present, room_dark)

    elif ch == 2:
        automation = False
        print("bye")
        break

    else:
        print("invalid choice")
```
