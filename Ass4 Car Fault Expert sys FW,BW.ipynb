```python
rules = [
    (["battery_weak", "lights_dim"], "battery_problem"),
    (["battery_problem", "clicking_sound"], "starter_problem"),
    (["engine_not_starting", "battery_problem"], "starting_fault"),

    (["engine_overheating", "coolant_low"], "cooling_fault"),
    (["brake_noise", "brake_soft"], "brake_fault"),
    (["tyre_pressure_low"], "tyre_problem"),
    (["fuel_empty"], "refuel_required"),
    (["ac_not_cooling"], "ac_fault"),

    (["clutch_hard", "gear_not_changing"], "clutch_problem"),
    (["engine_noise", "oil_low"], "oil_problem")
]


facts = {
    "battery_weak",
    "lights_dim",
    "clicking_sound",
    "engine_not_starting",
    "fuel_empty",
    "tyre_pressure_low",
    "clutch_hard",
    "gear_not_changing",
    "engine_noise"
}


def forward_chaining(KB, F):

    KnownFacts = set(F)
    AddedNewFact = True

    while AddedNewFact:

        AddedNewFact = False

        for antecedents, consequent in KB:

            if consequent not in KnownFacts:

                if set(antecedents).issubset(KnownFacts):

                    KnownFacts.add(consequent)

                    print("Rule fired:",
                          " AND ".join(antecedents),
                          "->", consequent)

                    AddedNewFact = True

    return KnownFacts


def backward_chaining(KB, KnownFacts, G):

    if G in KnownFacts:
        return True

    MatchingRules = []

    for antecedents, consequent in KB:

        if consequent == G:
            MatchingRules.append((antecedents, consequent))

    if not MatchingRules:
        return False

    for antecedents, consequent in MatchingRules:

        possible = True

        for A in antecedents:

            if not backward_chaining(KB, KnownFacts, A):
                possible = False
                break

        if possible:
            KnownFacts.add(G)
            return True

    return False


def display_rules():

    print("\nKnowledge Base:")

    print("battery_weak AND lights_dim -> battery_problem")
    print("battery_problem AND clicking_sound -> starter_problem")
    print("engine_not_starting AND battery_problem -> starting_fault")
    print("engine_overheating AND coolant_low -> cooling_fault")
    print("brake_noise AND brake_soft -> brake_fault")
    print("tyre_pressure_low -> tyre_problem")
    print("fuel_empty -> refuel_required")
    print("ac_not_cooling -> ac_fault")
    print("clutch_hard AND gear_not_changing -> clutch_problem")
    print("engine_noise AND oil_low -> oil_problem")


while True:

    print("\n CAR FAULT EXPERT SYSTEM ")
    print("1. Display Rules")
    print("2. Forward Chaining")
    print("3. Backward Chaining")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":

        display_rules()

    elif choice == "2":

        result = forward_chaining(rules, facts)

        print("All Facts:")

        for x in result:
            print(x)

    elif choice == "3":

        goal = input("Enter goal: ")

        if backward_chaining(rules, facts, goal):
            print("Goal proved:", goal)
        else:
            print("Goal cannot be proved:", goal)

    elif choice == "4":

        print("Exiting...")
        break

    else:

        print("Invalid choice")
```
