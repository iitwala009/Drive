```python
# Forward and Backward Chaining (Disease Prediction)

rules = {
    "flu": ["fever", "cough"],
    "cold": ["cough", "sneezing"],
    "infection": ["fever", "body_pain"],
    "pneumonia": ["fever", "cough", "breathing_problem"],
    "take_medicine": ["flu"],
    "take_rest": ["flu"]
}

known_facts = []
matching = []
ans = []
facts = []

for i in rules:
    for j in rules[i]:
        facts.append(j)

facts.append("flu")
facts.append("cold")
facts.append("infection")
facts.append("pneumonia")
facts.append("take_medicine")
facts.append("take_rest")



def forward(rules, known_facts):
    new_fact = True
    last_fact = known_facts[-1] if known_facts else None

    while new_fact:
        new_fact = False

        for consequent in rules:
            if all(antecedent in known_facts for antecedent in rules[consequent]):
                if consequent not in known_facts:
                    print(known_facts, "->", consequent)
                    known_facts.append(consequent)
                    last_fact = consequent
                    new_fact = True

    if last_fact is None:
        return "No conclusion"
    return last_fact




def backward(rules, known_facts, goal):
    global ans

    if goal in known_facts:
        if goal not in ans:
            ans.append(goal)
        return True

    if goal not in rules:
        return False

    for antecedent in rules[goal]:
        if not backward(rules, known_facts, antecedent):
            return False

    known_facts.append(goal)
    return True


# ------ Main program ------

print("Welcome to an intelligent disease detection system")
print("\n-----Menu-----")
print("1. Forward chaining")
print("2. Backward chaining")
print("3. Exit\n")

while True:
    option = input("Enter your choice: ")
    print("\n")

    if option == "1":

        print("1.fever\n2.cough\n3.sneezing\n4.body_pain\n5.breathing_problem\n6.flu\n7.cold\n8.infection\n9.pneumonia\n10.take_medicine\n11.take_rest")

        a = {
            "1": "fever",
            "2": "cough",
            "3": "sneezing",
            "4": "body_pain",
            "5": "breathing_problem",
            "6": "flu",
            "7": "cold",
            "8": "infection",
            "9": "pneumonia",
            "10": "take_medicine",
            "11": "take_rest"
        }

        m = int(input("Enter the no of known facts: "))
        known_facts = []

        while m > 0:
            i = input("Enter the no fact: ")

            if i not in a or a[i] not in facts:
                print("invalid fact give again")
            else:
                if a[i] not in known_facts:
                    known_facts.append(a[i])
                m -= 1

        print("ans :", forward(rules, known_facts), "\n")

    elif option == "2":

        print("1.fever\n2.cough\n3.sneezing\n4.body_pain\n5.breathing_problem\n6.flu\n7.cold\n8.infection\n9.pneumonia\n10.take_medicine\n11.take_rest")

        a = {
            "1": "fever",
            "2": "cough",
            "3": "sneezing",
            "4": "body_pain",
            "5": "breathing_problem",
            "6": "flu",
            "7": "cold",
            "8": "infection",
            "9": "pneumonia",
            "10": "take_medicine",
            "11": "take_rest"
        }

        while True:
            n = input("Enter the no Goal: ")

            if n not in a:
                print("invalid goal give again")
            else:
                goal = a[n]
                break

        known_facts = ["fever", "cough"]
        ans.clear()

        if backward(rules, known_facts, goal):
            print("Goal :", goal)
            print("Goal is proved")
            print("because :", ans)

        else:
            print("Goal is not proved")

    elif option == "3":
        print("Exiting......")
        break

    else:
        print("Invalid input")
```
