```python
# Plant Disease Detection System using Forward and Backward Chaining

kb = {
    "Stunted growth": ["White powder on leaves and stem", "Yield reduced"],
    "Fungal Infection": ["Stunted growth"],
    "Spray Medicine A on plant": ["Fungal Infection"]
}

initialf = set()

values = []
goal = "Spray Medicine A on plant"


def forwardchain():
    antecedent = set()
    print("Initial facts: ", initialf)

    knownfacts = initialf.copy()
    kflist = list(knownfacts)

    #print(knownfacts)

    addednewfact = True

    while addednewfact is True:
        addednewfact = False

        for rule in kb:
            value = kb.get(rule)
            antecedent.update(value)

            if rule not in knownfacts:
                if antecedent <= knownfacts:
                    knownfacts.add(rule)
                    kflist.append(rule)
                    addednewfact = True

    print("According to rules:", knownfacts)
    print("Final conclusion:", kflist.pop())


def backwardchain(kb, knownfacts, goal):

    if goal in knownfacts:
        return True

    if goal not in kb:
        return False

    antecedents = kb[goal]

    #allanteproven=True

    for a in antecedents:
        if not backwardchain(kb, knownfacts, a):
            #allanteproven=False
            return False

    knownfacts.add(goal)

    """if allanteproven==True:
        print("The goal :", goal, " is correct since -")
        print("Known facts:", knownfacts)"""

    return True


print("Welcome to Plant Disease Detection System. ")

while True:

    opno = int(input("Enter an option: 1. Forward Chaining  2. Backward Chaining  3. Exit"))

    if (opno == 1):

        print("Please answer the following questions: ")

        ans1 = input("Is there a white power on the leaves and stem of your plant? (Y/N)")

        while (ans1 != 'Y' and ans1 != 'N' and ans1 != 'y' and ans1 != 'n'):
            print("Invalid value.")
            ans1 = input("Is there a white power on the leaves and stem of your plant? (Y/N)")

        ans2 = input("Has your crop/fruit yield reduced: (Y/N)")

        while (ans2 != 'Y' and ans2 != 'N' and ans2 != 'y' and ans2 != 'n'):
            print("Invalid value.")
            ans2 = input("Has your crop/fruit yield reduced: (Y/N)")

        if ((ans2 == 'Y' or ans2 == 'y') and (ans1 == 'Y' or ans1 == 'y')):
            initialf.add("White powder on leaves and stem")
            initialf.add("Yield reduced")
            forwardchain()

        elif (((ans1 == 'Y' or ans1 == 'y') and (ans2 == 'N' or ans2 == 'n')) or
              ((ans2 == 'Y' or ans2 == 'y') and (ans1 == 'N' or ans1 == 'n'))):

            initialf.add("White powder on leaves and stem")
            initialf.add("Yield reduced")
            forwardchain()

        else:
            print("Your plant is healthy!")

    elif (opno == 2):

        ans3 = input("Enter your goal: ")

        if (ans3 == "Spray Medicine A on plant"):

            print("Goal is: ", goal)

            initialf.add("White powder on leaves and stem")
            initialf.add("Yield reduced")

            knownfacts = initialf.copy()

            if backwardchain(kb, knownfacts, goal):
                print("Goal is proven since -")
                print("Known facts:", knownfacts)

            else:
                print("Goal cannot be proven.")

        else:
            print("Sorry, this goal is unsupported. ")

    elif (opno == 3):

        print("Exiting...Thank you.")
        break

    else:
        print("Invalid Value. Try Again.")
```
